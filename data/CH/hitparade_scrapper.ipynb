```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import random

def scrape_hitparade():
    base_url = "https://hitparade.ch/charts/jahreshitparade/"
    start_year = 1968
    end_year = 2023
    
    all_songs = []

    for year in range(start_year, end_year + 1):
        print(f"Scrape data for year {year}...")
        
        url = f"{base_url}{year}"
        
        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
        }
        
        try:
            response = requests.get(url, headers=headers)
            
            if response.status_code != 200:
                print(f"Error in requesting year {year}. Status Code: {response.status_code}")
                continue
                
            soup = BeautifulSoup(response.text, 'html.parser')
            
            chart_entries = soup.find_all('div', class_='chartitem')
            
            for entry in chart_entries:
                try:
                    # Extraxt the rank
                    pos_div = entry.find('div', class_='chart_pos')
                    rank = pos_div.text.strip() if pos_div else "N/A"
                    
                    # Extrakt the artist and title
                    title_div = entry.find('div', class_='chart_title')
                    
                    if title_div and title_div.find('a'):
                        a_tag = title_div.find('a')
                        
                        # The artist is in the <b> tag
                        artist = a_tag.find('b').text.strip()
                        
                        # The title is in the end of the <a> tag after the <br> tag
                        title = str(a_tag.contents[-1]).strip()
                        
                        all_songs.append({
                            'Year': year,
                            'Rank': rank,
                            'Artist': artist,
                            'Title': title
                        })
                except Exception as inner_e:
                    continue
                    
        except Exception as e:
            print(f"An error occured in the year {year}: {e}")
            
        # Sleep 1-3 seconds between requests
        time.sleep(random.uniform(1.0, 3.0))

    # Convert to pandas dataframe
    df = pd.DataFrame(all_songs)
    
    # Save as CSV
    if not df.empty:
        df.to_csv('songs_original.csv', index=False, encoding='utf-8')
        print("\nScrapping done!")
        print(df.head())
    else:
        print("\nNo data found. Check the script.")

# Start script
if __name__ == "__main__":
    scrape_hitparade()
```
