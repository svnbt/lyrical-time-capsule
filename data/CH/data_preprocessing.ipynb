```python
import pandas as pd
import re
import emoji

def process_data():
    input_file = 'songs_lyrics.csv'
    output_file = 'songs_processed.csv'

    try:
        # Load data
        print("Loading data...")
        df = pd.read_csv(input_file)

        # Ensure Lyrics column is treated as strings
        df['Lyrics'] = df['Lyrics'].astype(str)

        original_len = len(df)

        # Drop entries containing emojis
        df = df[~df['Lyrics'].apply(lambda x: emoji.emoji_count(x) > 0)]

        # Drop entries containing numbering patterns
        df = df[~df['Lyrics'].str.contains(r'\b\d+\.\s', regex=True, na=False)]

        # Drop entries with too many quotation marks
        df = df[df['Lyrics'].str.count('"') <= 10]

        # Log how many rows were removed
        new_len = len(df)
        print(f"Dropped {original_len - new_len} rows containing emojis, numbering, or excessive quotes.")

        # Replace newlines with spaces
        df['Lyrics'] = df['Lyrics'].str.replace(r'[\n]+', ' ', regex=True)
        
        # If multiple spaces replace with one
        df['Lyrics'] = df['Lyrics'].str.replace(r'\s+', ' ', regex=True).str.strip()

        # Save data
        df.to_csv(output_file, index=False, encoding='utf-8')
        print(f"Success! Processed data saved to {output_file}")

    except Exception as e:
        print(f"An error has occurred: {e}")

if __name__ == "__main__":
    process_data()
```
